### Đề bài:
Tiền xử lý dữ liệu:
- Cắt bỏ các ký tự khoảng trắng thừa ở hai đầu tên mặt hàng mua (cột `Description`).
- Xóa các dòng dữ liệu không có số hóa đơn (cột `InvoiceNo`).
- Chuyển kiểu dữ liệu cột `InvoiceNo` về kiểu chuỗi (`str`).

```python
# Nguyễn Thành Trí - 6451071079
# Phần 2 - Bài 2 (Trang 37-38) - Câu 2: Tiền xử lý Description và InvoiceNo

import os
import pandas as pd
import numpy as np

print("Họ tên: Nguyễn Thành Trí - MSSV: 6451071079")

def get_data_path(rel_path):
    candidates = [
        rel_path,
        os.path.join("..", rel_path),
        os.path.join("..", "..", rel_path),
        os.path.join("datasets", os.path.basename(rel_path)),
        os.path.join("..", "datasets", os.path.basename(rel_path)),
        os.path.join("..", "..", "datasets", os.path.basename(rel_path))
    ]
    for p in candidates:
        if os.path.exists(p):
            return p
    raise FileNotFoundError(f"Không tìm thấy file: {rel_path}")

file_path = get_data_path("Online Retail.xlsx")
df = pd.read_excel(file_path)
print(f"Số dòng ban đầu: {len(df)}")

# 1. Cắt bỏ khoảng trắng thừa ở Description
df['Description'] = df['Description'].str.strip()

# 2. Xóa các dòng thiếu số hóa đơn (InvoiceNo rỗng/NaN)
df = df.dropna(subset=['InvoiceNo'])

# 3. Chuyển đổi InvoiceNo về kiểu chuỗi
df['InvoiceNo'] = df['InvoiceNo'].astype(str)

print(f"Số dòng sau khi làm sạch Description và lọc thiếu InvoiceNo: {len(df)}")
print("Kiểu dữ liệu cột InvoiceNo:", df['InvoiceNo'].dtype)
display(df.head())
```

### Nhận xét & Đánh giá (Câu 2):
1. Cắt bỏ khoảng trắng thừa (`strip()`) ở `Description` giúp chuẩn hóa tên mặt hàng, đảm bảo các mặt hàng giống nhau không bị tách thành hai tên phân biệt chỉ vì dấu cách dư thừa ở đầu/cuối chuỗi.
2. Việc ép kiểu `InvoiceNo` về kiểu chuỗi (`str`) giúp tránh tình trạng số hóa đơn bị coi là số nguyên, đồng thời tạo điều kiện thuận lợi để kiểm tra các ký tự tiền tố như hóa đơn hủy ('C') ở câu tiếp theo.
